```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
delivery = pd.read_csv('deliveries.csv')
```

```python
delivery
```

```python
deth_over=delivery[delivery['over'] > 15]
```

```python
all_batsman=deth_over.groupby('batsman')['batsman_runs'].count()
x=all_batsman>200
all_batsman[x]
```

```python
all_batsman[x].shape
```

```python
batsman_list=all_batsman[x].index.tolist()
```

```python
# Runs scored by all these 43 batsman
# Balls played by these 43 batsman

final=delivery[delivery['batsman'].isin(batsman_list)]
```

```python
runs=final.groupby('batsman')['batsman_runs'].sum()
```

```python
balls=final.groupby('batsman')['batsman_runs'].count()
```

```python
sr=(runs/balls)*100
```
